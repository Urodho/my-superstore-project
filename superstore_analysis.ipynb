```python
# 1. Missing values
print(orders.isnull().sum())
print(returns.isnull().sum())
print(people.isnull().sum())

# 2. Duplicates
print(orders.duplicated().sum())
print(returns.duplicated().sum())
print(people.duplicated().sum())

# 3. Data types
print(orders.dtypes)
```

```python
orders["Order Date"] = pd.to_datetime(orders["Order Date"], dayfirst=True)
orders["Ship Date"] = pd.to_datetime(orders["Ship Date"], dayfirst=True)
print(orders[["Order Date", "Ship Date"]].head())
```

Missing values: Postal Code(column). I left/handled them because were missing from 41,296 rows. Duplicate rows: 0. Dates were converted to datetime so I can calculate time differences.

Create Calculated Columns

```python
orders["Shipping Duration"] = (orders["Ship Date"] - orders["Order Date"]).dt.days

orders["Profit Margin"] = orders["Profit"] / orders["Sales"].replace(0, np.nan) * 100

orders[["Order Date", "Ship Date", "Shipping Duration", "Sales", "Profit", "Profit Margin"]].head()
```

I created Shipping Duration (days between order and shipping) and Profit Margin (Profit ÷ Sales × 100). Sales of zero were handled to avoid division errors.

Basic Business Analysis

```python
total_sales = orders["Sales"].sum()
total_profit = orders["Profit"].sum()
total_quantity = orders["Quantity"].sum()
avg_discount = orders["Discount"].mean()
total_shipping = orders["Shipping Cost"].sum()
avg_ship_days = orders["Shipping Duration"].mean()
avg_margin = orders["Profit Margin"].mean()
overall_margin = total_profit / total_sales * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Average Discount:", round(avg_discount * 100, 2), "%")
print("Total Shipping Cost:", round(total_shipping, 2))
print("Average Shipping Duration:", round(avg_ship_days, 2), "days")
print("Average Profit Margin (row-level):", round(avg_margin, 2), "%")
print("Overall Profit Margin (total profit / total sales):", round(overall_margin, 2), "%")
```

The company made total sales of 12642501.91 and total profit of 1467457.29, an overall margin of 11.61%. The average discount was 4.74% and orders took on average 3.97 days to ship

MERGING THE DATA SETS

```python
print(returns.columns)
print(returns.head())
print(returns["Order ID"].duplicated().sum())
```

```python
# Keep one row per returned order, with a flag column
returns_clean = returns[["Order ID"]].drop_duplicates()
returns_clean["Returned"] = "Yes"

# Left merge: keep ALL orders, attach return info where it exists
orders_merged = orders.merge(returns_clean, on="Order ID", how="left")

# Orders with no match get NaN, so mark them "No"
orders_merged["Returned"] = orders_merged["Returned"].fillna("No")

# Safety check: row count must be unchanged
print(orders.shape[0], orders_merged.shape[0])
print(orders_merged["Returned"].value_counts())
```

```python
print(people.columns)
print(people.head())
print(orders["Region"].nunique(), people["Region"].nunique())
```

```python
orders_merged = orders_merged.merge(people, on="Region", how="left")
print(orders.shape[0], orders_merged.shape[0])
print(orders_merged["Person"].isnull().sum())
```

I used a left merge on Order ID so all orders were kept and returned orders were flagged. I removed duplicate Order IDs from Returns first to avoid inflating rows. The row count stayed the same after merging, confirming the join was correct. (add:) 384 orders were returned.

GroupBy Analysis

From now on i use orders_merged, because it has all the original columns plus Returned

```python
def summarize(group_col):
    result = orders_merged.groupby(group_col).agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Quantity_Sold=("Quantity", "sum"),
        Avg_Sales=("Sales", "mean"),
        Avg_Profit=("Profit", "mean"),
        Num_Orders=("Order ID", "nunique")
    ).round(2)
    return result

market_summary = summarize("Market")
region_summary = summarize("Region")
country_summary = summarize("Country")
category_summary = summarize("Category")
subcat_summary = summarize("Sub-Category")
segment_summary = summarize("Segment")

market_summary
```

Pivot
-A pivot table reorganizes data into a grid. One variable goes down the rows, another across the columns, and each cell holds a calculation (like total profit). It's the same idea as an Excel pivot table, and it lets you compare two dimensions at once.t Table


```python
pivot_profit = pd.pivot_table(
    orders_merged,
    values="Profit",
    index="Market",
    columns="Category",
    aggfunc="sum",
    margins=True,
    margins_name="Total",
    fill_value=0
).round(2)

pivot_profit
```

A second pivot for sales, so you can compare:

```python
pivot_sales = pd.pivot_table(
    orders_merged,
    values="Sales",
    index="Segment",
    columns="Market",
    aggfunc="sum",
    margins=True,
    margins_name="Total"
).round(2)

pivot_sales
```

NumPy, Profit Classification

NumPy lets us apply "if this, then that" rules to a whole column at once.

np.where(condition, value_if_true, value_if_false) handles one yes/no rule.
np.select(conditions, choices, default) handles several rules. It checks the conditions in order and uses the first one that is True.

```python
orders_merged["Profit"].describe()
```

The cutoffs below (50 and 200) are my suggestion, not a rule. If your describe() output shows very different numbers (look at the 50% and 75% rows), change them, and say in your notebook that you chose them.

```python
conditions = [
    orders_merged["Profit"] < 0,
    orders_merged["Profit"] < 50,
    orders_merged["Profit"] < 200
]

choices = ["Loss", "Low Profit", "Medium Profit"]

orders_merged["Profit Class"] = np.select(conditions, choices, default="High Profit")

orders_merged["Profit Class"].value_counts()
```

Simple np.where example (one rule only):

```python
orders_merged["Profitable"] = np.where(orders_merged["Profit"] > 0, "Yes", "No")
orders_merged["Profitable"].value_counts()
```

discounting and profit

```python
orders_merged["Discount Band"] = pd.cut(
    orders_merged["Discount"],
    bins=[-0.01, 0, 0.2, 0.4, 1.0],
    labels=["No discount", "1-20%", "21-40%", "Over 40%"]
)

discount_analysis = orders_merged.groupby("Discount Band", observed=True).agg(
    Avg_Profit=("Profit", "mean"),
    Total_Profit=("Profit", "sum"),
    Num_Rows=("Profit", "count")
).round(2)

discount_analysis
```

```python
orders_merged["Discount"].corr(orders_merged["Profit"])
```

I used np.select to classify each order line by profit. 21-40% rows were losses and 1-20% were high profit. Discounts and profit have a correlation of moderate negative. Orders with discounts over 40% had an average profit of -90.13, compared to 61.04 with no discount.

Final Business Summary.

Business Summary: Global Superstore Analysis

What I analyzed: I analyzed about 51,290 order lines from 2014 to 2017 using Pandas. I cleaned the data, calculated shipping duration and profit margin, merged Orders with Returns, and analyzed sales and profit by market, region, category, sub-category, segment and time.

Key findings:

Total sales were about 12.64 million and total profit about 1.47 million, an overall profit margin of about 11.6%.
Europe is the most profitable market, and Western Europe the most profitable region. Asia Pacific has the highest sales.
Technology is the most profitable category, and Copiers the most profitable sub-category.
The Consumer segment generates the most sales.
Sales grew about 90% and profit about doubled from 2014 to 2017.
Discounts hurt profit. Orders with no discount averaged a profit of 61.04 per line, while those with discounts over 40% averaged a loss of 90.13.

Problems discovered:

24.46% of order lines lost money.
Discounts above 20% lost about 814,700 in total.
Tables loses money (-64,083) despite high sales.
Turkey, Nigeria, the Netherlands, Honduras and Pakistan are loss-making countries.
7.66% of orders were returned.

Opportunities:

Invest further in Technology, especially Copiers, and in the strong European market.
Grow the Consumer segment, which brings the most sales.
Use faster shipping options as a selling point where customers value speed.

Recommendations:

Limit discounts to 20% or below, because deeper discounts push average profit below zero.
Review pricing and costs for Tables, which sell well but lose money.
Investigate the loss-making countries, starting with Turkey and Nigeria, for pricing, shipping or discounting problems.
Study why 7.66% of orders are returned, and look for patterns by product or region.
Keep expanding in Europe and Technology, which are the main profit drivers.
